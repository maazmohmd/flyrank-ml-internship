## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

```python
Research question: Which existing content pages should a content reviewer inspect first for possible refresh or other content action, based on observable search, content, and engagement signals?
Unit of analysis: One content page.
Decision: Which pages should be reviewed first.
Action: A content reviewer can examine the highest-ranked pages and decide whether they should be refreshed, expanded, monitored, or left unchanged.
Cost of a wrong recommendation: A wrong recommendation could waste time by sending reviewers to pages that do not need attention.
Missing an important page could also mean that a useful content opportunity is not reviewed.
Why data/ML can help: There are many signals such as clicks, impressions, position, CTR, page age, freshness, and engagement.
Data analysis can help find useful patterns across these signals and prioritize pages more efficiently.
```

```python
The starter dataset contains 30,000 content pages from 32 clients.
Around 16,262 pages, or 54.2%, show a downward trend according to the starter dataset's trend definition.
This means there are many possible pages that could require investigation.
Since reviewing every page would take a lot of time, prioritizing which pages should be reviewed first could be useful.
```

```python
import pandas as pd

df = pd.read_csv("../../data/raw/content_refresh_anonymized.csv")

print("Number of pages:", len(df))
print("Number of clients:", df["client_id"].nunique())

declining = (df["trend_direction"] == "down").sum()
declining_pct = (df["trend_direction"] == "down").mean() * 100

print("Pages with downward trend:", declining)
print(f"Percentage with downward trend: {declining_pct:.1f}%")

```
